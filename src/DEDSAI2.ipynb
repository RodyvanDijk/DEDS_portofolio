```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
from sklearn import metrics
import matplotlib.pyplot as plt
```

```python
import sqlite3
try:
    sqliteConnection = sqlite3.connect('go_sales_train.sqlite')
    cursor = sqliteConnection.cursor()
    print('Db created and connected succesfull')
except sqlite3.Error as error:
    print('Error occured: ', error)
```

```python
sales_returnedItem = pd.read_sql("SELECT * FROM returned_item", sqliteConnection)
sales_orderDetails = pd.read_sql("SELECT * FROM order_details", sqliteConnection)

table = pd.merge(sales_returnedItem, sales_orderDetails, left_on='ORDER_DETAIL_CODE', right_on='ORDER_DETAIL_CODE')

table
```

```python
table = table.loc[:, ['RETURN_REASON_CODE', 'PRODUCT_NUMBER', 'QUANTITY', 'RETURN_QUANTITY']]
table
```

```python
X = table.drop('RETURN_REASON_CODE', axis = 1)
y = table.loc[:, ['RETURN_REASON_CODE']]
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.15,
    random_state=42
)
```

```python
list(X.columns)
```

```python
dtree = DecisionTreeClassifier(max_depth = 3)
dtree = dtree.fit(X_train, y_train)
tree.plot_tree(dtree, feature_names = X.columns)
plt.show()
```

```python
predicted_df = pd.DataFrame(dtree.predict(X_test))
predicted_df = predicted_df.rename(columns = {0 : 'Predicted_Return_Reason'})
model_results_frame = pd.concat([y_test.reset_index()['RETURN_REASON_CODE'], predicted_df], axis = 1)
model_results_frame
```

```python
confusion_matrix = metrics.confusion_matrix(model_results_frame['RETURN_REASON_CODE'], model_results_frame['Predicted_Return_Reason'])
cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = confusion_matrix, display_labels = [1,2,3,4,5])

cm_display.plot()
plt.show()
```

```python
metrics.accuracy_score(model_results_frame['RETURN_REASON_CODE'], model_results_frame['Predicted_Return_Reason'])
```
